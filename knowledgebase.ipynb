```python
facts = {"bird", "has_wings"}
rules = [("bird", "can_fly"), ("has_wings", "can_move")]

knowledge_base = {"facts": facts, "rules": rules}
print("Facts:", knowledge_base["facts"])
for condition, result in rules:
    if condition in facts:
        print(condition, "->", result)
```
