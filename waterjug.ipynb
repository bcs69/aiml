```python
from collections import deque

capacity1 = 4
capacity2 = 3
goal = 2
queue = deque([((0, 0), [])])
visited = {(0, 0)}

while queue:
    (a, b), path = queue.popleft()
    if a == goal or b == goal:
        print("Steps:")
        for step in path + [(a, b)]:
            print(step)
        break

    states = [
        (capacity1, b),
        (a, capacity2),
        (0, b),
        (a, 0),
        (a - min(a, capacity2 - b), b + min(a, capacity2 - b)),
        (a + min(b, capacity1 - a), b - min(b, capacity1 - a))
    ]
    for state in states:
        if state not in visited:
            visited.add(state)
            queue.append((state, path + [(a, b)]))
```
