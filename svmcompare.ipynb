```python
from sklearn.svm import SVC

X = [[1,2],[2,3],[3,1],[4,2],[5,5],[6,6],[7,4],[8,5]]
y = [0,0,0,0,1,1,1,1]

linear = SVC(kernel="linear")
rbf = SVC(kernel="rbf", gamma="scale")
linear.fit(X, y)
rbf.fit(X, y)
print("Linear SVM:", linear.predict([[4,4]])[0])
print("RBF SVM:", rbf.predict([[4,4]])[0])
```
