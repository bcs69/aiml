```python
n = 4
board = [-1] * n

def safe(row, col):
    for previous in range(row):
        if board[previous] == col or abs(board[previous] - col) == abs(previous - row):
            return False
    return True

def solve(row):
    if row == n:
        return True
    for col in range(n):
        if safe(row, col):
            board[row] = col
            if solve(row + 1):
                return True
            board[row] = -1
    return False

solve(0)
for row in board:
    print(" ".join("Q" if col == row else "." for col in board))
```
