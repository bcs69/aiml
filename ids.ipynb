```python
def dls(node, goal, depth, graph, path):
    if node == goal:
        return path
    if depth == 0:
        return None
    for neighbor in graph[node]:
        result = dls(neighbor, goal, depth - 1, graph, path + [neighbor])
        if result:
            return result
    return None

graph = {"A": ["B", "C"], "B": ["D", "E"], "C": ["F"], "D": [], "E": [], "F": []}
for limit in range(4):
    result = dls("A", "F", limit, graph, ["A"])
    if result:
        print("Depth limit:", limit)
        print("Path:", result)
        break
```
