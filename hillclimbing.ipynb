```python
def f(x):
    return -x * x + 4 * x

x = 0
while True:
    current = f(x)
    left = f(x - 1)
    right = f(x + 1)
    if left > current:
        x -= 1
    elif right > current:
        x += 1
    else:
        break

print("Maximum at x =", x)
print("Maximum value =", f(x))
```
