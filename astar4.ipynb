```python
import heapq

start = (1, 2, 3, 4, 0, 5, 6, 7, 8)
goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)

moves = {0: [1, 3], 1: [0, 2, 4], 2: [1, 5], 3: [0, 4, 6], 4: [1, 3, 5, 7], 5: [2, 4, 8], 6: [3, 7], 7: [4, 6, 8], 8: [5, 7]}

def heuristic(state):
    return sum(value != 0 and value != goal[i] for i, value in enumerate(state))

queue = [(heuristic(start), 0, start, [start])]
visited = {start}
while queue:
    f, cost, state, path = heapq.heappop(queue)
    if state == goal:
        print("Moves:", len(path) - 1)
        for item in path:
            print(item)
        break
    zero = state.index(0)
    for pos in moves[zero]:
        new_state = list(state)
        new_state[zero], new_state[pos] = new_state[pos], new_state[zero]
        new_state = tuple(new_state)
        if new_state not in visited:
            visited.add(new_state)
            new_cost = cost + 1
            heapq.heappush(queue, (new_cost + heuristic(new_state), new_cost, new_state, path + [new_state]))
```
