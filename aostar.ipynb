```python
graph = {
    "A": [[("B", 2)], [("C", 3)]],
    "B": [[("D", 1), ("E", 4)]],
    "C": [[("F", 2)], [("G", 5)]],
    "D": [[("H", 2)], [("I", 3)]],
    "E": [[("J", 1)]]
}
goals = {"F": 0, "G": 0, "H": 0, "I": 0, "J": 0}

from functools import lru_cache

@lru_cache(None)
def cost(node):
    if node in goals:
        return goals[node], [node]
    options = []
    for option in graph[node]:
        total = 0
        path = [node]
        for child, edge in option:
            child_cost, child_path = cost(child)
            total += edge + child_cost
            path += child_path
        options.append((total, path))
    return min(options, key=lambda x: x[0])

value, path = cost("A")
print("AO* cost:", value)
print("Solution path:", path)
```
