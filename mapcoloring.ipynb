```python
graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "D"],
    "D": ["B", "C"]
}
colors = ["Red", "Green", "Blue"]
assignment = {}

def color_map(node):
    if node == len(graph):
        return True
    city = list(graph)[node]
    for color in colors:
        if all(assignment.get(neighbor) != color for neighbor in graph[city]):
            assignment[city] = color
            if color_map(node + 1):
                return True
            del assignment[city]
    return False

color_map(0)
print(assignment)
```
