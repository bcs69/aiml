```python
from sklearn.neural_network import MLPRegressor

X = [[1],[2],[3],[4],[5],[6]]
y = [2,4,6,8,10,12]

model = MLPRegressor(hidden_layer_sizes=(5,), learning_rate_init=0.01, max_iter=2000, random_state=42)
model.fit(X, y)
print("Predicted value for 7:", round(model.predict([[7]])[0], 2))
```
