```python
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [], "E": [], "F": []
}

def dls(node, goal, depth, path):
    if node == goal:
        return path
    if depth == 0:
        return None
    for neighbor in graph[node]:
        result = dls(neighbor, goal, depth - 1, path + [neighbor])
        if result:
            return result
    return None

print("Path:", dls("A", "F", 2, ["A"]))
```
