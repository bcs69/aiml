```python
graph = {
    1: [2, 3],
    2: [4],
    3: [5],
    4: [6],
    5: [6],
    6: []
}

visited = set()
order = []
def dfs(node):
    visited.add(node)
    order.append(node)
    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(neighbor)

dfs(1)
print("DFS from 1:", order)
```
