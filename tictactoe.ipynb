```python
def winner(board):
    lines = [(0,1,2),(3,4,5),(6,7,8),(0,3,6),(1,4,7),(2,5,8),(0,4,8),(2,4,6)]
    for a, b, c in lines:
        if board[a] == board[b] == board[c] and board[a] != " ":
            return board[a]
    return None

def minimax(board, maximizing):
    result = winner(board)
    if result == "O":
        return 1
    if result == "X":
        return -1
    if " " not in board:
        return 0
    scores = []
    for i in range(9):
        if board[i] == " ":
            board[i] = "O" if maximizing else "X"
            scores.append(minimax(board, not maximizing))
            board[i] = " "
    return max(scores) if maximizing else min(scores)

def best_move(board):
    best_score = -2
    move = None
    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            score = minimax(board, False)
            board[i] = " "
            if score > best_score:
                best_score = score
                move = i
    return move

board = ["X", "O", "X", " ", "O", " ", " ", " ", " "]
move = best_move(board)
print("Best computer move:", move + 1)
```
