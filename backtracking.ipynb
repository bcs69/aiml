```python
graph = {
    1: [2, 3, 4, 5],
    2: [1, 3, 4, 5],
    3: [1, 2, 4, 5],
    4: [1, 2, 3, 5],
    5: [1, 2, 3, 4]
}
colors = ["Red", "Green", "Blue", "Yellow", "Purple"]
assignment = {}

nodes = list(graph)
def solve(index):
    if index == len(nodes):
        return True
    node = nodes[index]
    for color in colors:
        if all(assignment.get(neighbor) != color for neighbor in graph[node]):
            assignment[node] = color
            if solve(index + 1):
                return True
            del assignment[node]
    return False

solve(0)
print(assignment)
```
