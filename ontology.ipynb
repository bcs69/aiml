```python
ontology = {
    "Animal": {"subclasses": ["Bird", "Mammal"]},
    "Bird": {"subclasses": ["Parrot"]},
    "Mammal": {"subclasses": ["Dog"]}
}
relations = [("Dog", "has", "four legs"), ("Parrot", "can", "fly")]

for parent, data in ontology.items():
    print(parent, "->", data["subclasses"])
for relation in relations:
    print(relation)
```
