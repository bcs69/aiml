```python
import heapq

graph = {
    "A": {"B": 2, "C": 4},
    "B": {"D": 3, "E": 5},
    "C": {"E": 1, "F": 1},
    "D": {"F": 2},
    "E": {"F": 1},
    "F": {}
}
heuristic = {"A": 6, "B": 5, "C": 3, "D": 2, "E": 1, "F": 0}

queue = [(heuristic["A"], "A", ["A"])]
visited = set()
while queue:
    _, node, path = heapq.heappop(queue)
    if node in visited:
        continue
    visited.add(node)
    if node == "F":
        print("Path:", path)
        break
    for neighbor in graph[node]:
        if neighbor not in visited:
            heapq.heappush(queue, (heuristic[neighbor], neighbor, path + [neighbor]))
```
