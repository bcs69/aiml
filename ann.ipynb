```python
from sklearn.neural_network import MLPClassifier

X = [[3, 1.5], [2, 1], [4, 1.5], [3, 4], [3.5, 0.5], [2, 0.5], [5.5, 1], [1, 1]]
y = [0, 1, 0, 1, 0, 1, 1, 0]

for learning_rate in [0.1, 0.01]:
    for epochs in [20, 100]:
        model = MLPClassifier(hidden_layer_sizes=(5,), learning_rate_init=learning_rate, max_iter=epochs, random_state=42)
        model.fit(X, y)
        print("Learning rate:", learning_rate, "Epochs:", epochs, "Accuracy:", model.score(X, y))
```
