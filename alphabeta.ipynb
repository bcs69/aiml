```python
def minimax_alpha_beta(node, depth, maximizing, alpha, beta):
    if isinstance(node, int):
        return node
    if maximizing:
        value = float("-inf")
        for child in node:
            value = max(value, minimax_alpha_beta(child, depth + 1, False, alpha, beta))
            alpha = max(alpha, value)
            if alpha >= beta:
                break
        return value
    value = float("inf")
    for child in node:
        value = min(value, minimax_alpha_beta(child, depth + 1, True, alpha, beta))
        beta = min(beta, value)
        if alpha >= beta:
            break
    return value

tree = [
    [3, 5],
    [2, 9],
    [0, 7]
]
print("Best value:", minimax_alpha_beta(tree, 0, True, float("-inf"), float("inf")))
```
