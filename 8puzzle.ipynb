```python
import heapq

start = (1, 2, 3, 4, 0, 5, 6, 7, 8)
goal = (1, 2, 3, 4, 5, 6, 7, 8, 0)
moves = {0: [1, 3], 1: [0, 2, 4], 2: [1, 5], 3: [0, 4, 6], 4: [1, 3, 5, 7], 5: [2, 4, 8], 6: [3, 7], 7: [4, 6, 8], 8: [5, 7]}

def h(state):
    return sum(value != 0 and value != goal[i] for i, value in enumerate(state))

queue = [(h(start), 0, start)]
seen = {start}
while queue:
    f, cost, state = heapq.heappop(queue)
    if state == goal:
        print("Solved in", cost, "moves")
        break
    zero = state.index(0)
    for pos in moves[zero]:
        new = list(state)
        new[zero], new[pos] = new[pos], new[zero]
        new = tuple(new)
        if new not in seen:
            seen.add(new)
            heapq.heappush(queue, (cost + 1 + h(new), cost + 1, new))
```
