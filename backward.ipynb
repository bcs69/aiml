```python
rules = {
    "warm": {"sunny"},
    "go_outside": {"warm"},
    "happy": {"sunny", "warm"}
}
facts = {"sunny"}

def prove(goal, seen=None):
    seen = set() if seen is None else seen
    if goal in facts:
        return True
    if goal in seen or goal not in rules:
        return False
    seen.add(goal)
    return all(prove(condition, seen) for condition in rules[goal])

print("happy:", prove("happy"))
print("go_outside:", prove("go_outside"))
```
