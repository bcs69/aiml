```python
facts = {"sunny"}
rules = [
    ({"sunny"}, "warm"),
    ({"warm"}, "go_outside"),
    ({"warm", "sunny"}, "happy")
]

changed = True
while changed:
    changed = False
    for conditions, result in rules:
        if conditions.issubset(facts) and result not in facts:
            facts.add(result)
            changed = True

print("Facts after forward chaining:", sorted(facts))
```
