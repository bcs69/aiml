```python
graph = {
    "A": ["B", "C", "E"],
    "B": ["D"],
    "C": ["F"],
    "D": ["E"],
    "E": ["F"],
    "F": []
}

visited = set()
order = []

def dfs(node):
    visited.add(node)
    order.append(node)
    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(neighbor)

dfs("A")
print("DFS traversal:", order)
```
