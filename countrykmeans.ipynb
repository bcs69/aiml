```python
import os
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

columns = ["country nm", "child_mortality", "exports", "health", "imports", "income", "inflation", "life_expectancy", "total_fer", "gdp"]
if os.path.exists("country_data.csv"):
    data = pd.read_csv("country_data.csv")
else:
    data = pd.DataFrame([
        ["A", 90, 20, 5, 40, 1000, 8, 55, 4.5, 500],
        ["B", 60, 30, 6, 35, 2000, 6, 60, 3.8, 900],
        ["C", 30, 40, 8, 30, 4000, 4, 68, 2.8, 1800],
        ["D", 15, 50, 10, 25, 8000, 3, 72, 2.0, 3500],
        ["E", 75, 25, 5, 45, 1500, 7, 58, 4.0, 700],
        ["F", 20, 45, 9, 28, 7000, 3, 70, 2.2, 3000],
        ["G", 50, 35, 7, 32, 3000, 5, 65, 3.0, 1500],
        ["H", 25, 48, 9, 27, 9000, 2, 74, 1.8, 4200],
        ["I", 70, 28, 6, 38, 1800, 6, 59, 3.6, 800],
        ["J", 10, 55, 11, 22, 12000, 2, 76, 1.6, 5200]
    ], columns=columns)

X = data.drop(columns=["country nm"])
scaled = StandardScaler().fit_transform(X)

inertia = []
for k in range(1, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(scaled)
    inertia.append(model.inertia_)

plt.plot(range(1, 11), inertia, marker="o")
plt.xlabel("K")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.show()

model = KMeans(n_clusters=3, random_state=42, n_init=10)
data["cluster"] = model.fit_predict(scaled)
print(data[["country nm", "cluster"]])
```
