```python
def minimax(node, maximizing):
    if isinstance(node, int):
        return node
    values = [minimax(child, not maximizing) for child in node]
    return max(values) if maximizing else min(values)

def alphabeta(node, maximizing, alpha=float("-inf"), beta=float("inf")):
    if isinstance(node, int):
        return node, 1
    visited = 1
    if maximizing:
        value = float("-inf")
        for child in node:
            child_value, count = alphabeta(child, False, alpha, beta)
            visited += count
            value = max(value, child_value)
            alpha = max(alpha, value)
            if alpha >= beta:
                break
        return value, visited
    value = float("inf")
    for child in node:
        child_value, count = alphabeta(child, True, alpha, beta)
        visited += count
        value = min(value, child_value)
        beta = min(beta, value)
        if alpha >= beta:
            break
    return value, visited

tree = [[3, 5], [2, 9], [0, 7]]
print("Minimax value:", minimax(tree, True))
value, visited = alphabeta(tree, True)
print("Alpha-Beta value:", value)
print("Nodes visited by Alpha-Beta:", visited)
```
