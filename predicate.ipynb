```python
people = ["Amit", "Neha", "Rahul"]
students = {"Amit", "Neha"}
smart = {"Amit", "Rahul"}

print("Student(Amit):", "Amit" in students)
print("Student(Neha):", "Neha" in students)
print("There exists a smart student:", any(person in smart and person in students for person in people))
print("All students are smart:", all(person in smart for person in students))
```
