```python
import heapq

graph = {
    "A": {"B": 2, "C": 4},
    "B": {"D": 2, "E": 5},
    "C": {"F": 3},
    "D": {"G": 3},
    "E": {"G": 1},
    "F": {"G": 2},
    "G": {}
}
heuristic = {"A": 6, "B": 5, "C": 4, "D": 3, "E": 2, "F": 2, "G": 0}

def astar(start, goal):
    pq = [(heuristic[start], 0, start, [start])]
    best = {start: 0}
    while pq:
        f, cost, node, path = heapq.heappop(pq)
        if node == goal:
            return path, cost
        for next_node, edge in graph[node].items():
            new_cost = cost + edge
            if new_cost < best.get(next_node, float("inf")):
                best[next_node] = new_cost
                heapq.heappush(pq, (new_cost + heuristic[next_node], new_cost, next_node, path + [next_node]))

path, cost = astar("A", "G")
print("Shortest path:", path)
print("Cost:", cost)
```
